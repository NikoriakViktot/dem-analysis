```python
from pathlib import Path
import rasterio
import numpy as np

# === 🔧 Налаштування ===
base_folder = Path("/mnt/c/Users/5302/OneDrive/PhD/paper_DEM_artickle/data/DATA_UTM/dem/")
hand_folder = base_folder / "hand_outputs"
output_folder = base_folder / "flood_scenarios"
output_folder.mkdir(parents=True, exist_ok=True)

# 💧 Порогові рівні затоплення (м)
levels = list(range(1, 11))  # 1 м до 10 м

# 🔍 Знаходимо всі HAND-файли з threshold=2000
hand_files = sorted(hand_folder.glob("*_hand_2000.tif"))
print(f"🔎 Знайдено HAND-файлів: {len(hand_files)}")

for hand_path in hand_files:
    print(f"\n📂 Обробка: {hand_path.name}")

    with rasterio.open(hand_path) as src:
        hand = src.read(1)
        meta = src.meta.copy()

    # Маскуємо некоректні значення
    hand = np.where((hand >= 0) & np.isfinite(hand), hand, np.nan)

    for level in levels:
        flood_depth = np.where(hand <= level, level - hand, 0).astype("float32")

        out_name = f"{hand_path.stem}_flood_{level}m.tif"
        out_path = output_folder / out_name

        meta.update(dtype="float32", count=1, nodata=0.0, compress="lzw")

        with rasterio.open(out_path, "w", **meta) as dst:
            dst.write(flood_depth, 1)

        print(f"✅ Збережено: {out_name}")

print(f"\n🎉 Усі сценарії затоплення збережено в: {output_folder}")
```
