# DownslopeDistanceToStream — WhiteboxTools

**DownslopeDistanceToStream** — це інструмент для розрахунку відстані від кожної клітинки DEM до найближчого потоку (stream), виміряної вздовж потоку стікання (flowpath). Відстань розраховується або за D8, або за D-infinity flow алгоритмом.

---

## 🔎 **Для чого використовується?**
- Виявлення просторового патерну відстаней до річки для моделювання floodplain, аналізу рельєфу, побудови геоморфометричних ознак.
- Класична ознака для машинного навчання, аналізу ерозії, екології та гідрології.

---

## ⚙️ **Вхідні дані:**
- **DEM**: цифрова модель рельєфу, попередньо оброблена (`BreachDepressions`).
- **Streams**: raster з річками (всі не-нульові клітинки — річки), створений за допомогою порогу накопичення стоку (`D8FlowAccumulation` + threshold).
- **(Опційно) D8 Pointer**: автоматично використовується.

---

## 🧑‍💻 **Параметри (Python):**

```python
wbt.downslope_distance_to_stream(
    dem,          # шлях до DEM (raster .tif)
    streams,      # шлях до raster streams (не-нульові значення — річки)
    output,       # шлях до вихідного raster (відстань у метрах)
    dinf=False    # (опціонально) використовувати D-infinity алгоритм?
)
````

## 💻 **Приклад командного рядка (CLI):**

```bash
whitebox_tools -r=DownslopeDistanceToStream \
--dem=dem.tif \
--streams=streams.tif \
-o=distance_to_stream.tif
```

* **--dinf** — якщо хочеш використовувати D-infinity алгоритм.

---

## 🟢 **Короткий workflow для floodplain mapping:**

1. **BreachDepressions**: обробка DEM
2. **D8FlowAccumulation**: накопичення стоку
3. **Threshold**: створення stream raster
4. **DownslopeDistanceToStream**: розрахунок відстані до річки

`````python
from whitebox.whitebox_tools import WhiteboxTools
from pathlib import Path
`````

`````python
output_folder = Path("/mnt/c/Users/5302/OneDrive/PhD/paper_DEM_artickle/data/DATA_UTM/dem/distance_to_stream/")
output_folder.mkdir(exist_ok=True)
`````

`````python
stream_folder = Path("/mnt/c/Users/5302/OneDrive/PhD/paper_DEM_artickle/data/DATA_UTM/dem/stream_raster")
pointer_folder = Path("/mnt/c/Users/5302/OneDrive/PhD/paper_DEM_artickle/data/DATA_UTM/dem/flow_outputs")
dem_folder = Path("/mnt/c/Users/5302/OneDrive/PhD/paper_DEM_artickle/data/DATA_UTM/dem/breached_dems")
`````

`````python
# Для кожного DEM
wbt = WhiteboxTools()
wbt.set_working_dir(str(output_folder))  # output_folder для distance rasters

for dem_path in sorted(dem_folder.glob("*_breached.tif")):
    name = dem_path.stem.replace("_breached", "")

    pointer_path = pointer_folder / f"{name}_d8_pointer.tif"
    stream_path = stream_folder / f"{name}_stream_raster_2000.tif"
    out_distance = output_folder / f"{name}_distance_to_stream.tif"

    print(f"🚰 Відстань до річки для DEM: {name}")
    wbt.downslope_distance_to_stream(
        dem=str(dem_path),
        streams=str(stream_path),
        output=str(out_distance)
    )

    print(f"✅ Збережено: {out_distance}")

print("\n🎉 Усі distance-to-stream rasters готові!")
`````
