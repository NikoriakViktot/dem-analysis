```python
from whitebox.whitebox_tools import WhiteboxTools
from pathlib import Path

# Вхідні та вихідні папки
dem_folder = Path("/mnt/c/Users/5302/OneDrive/PhD/paper_DEM_artickle/data/DATA_UTM/dem/breached_dems")
output_folder = dem_folder.parent / "flow_TWI"
output_folder.mkdir(exist_ok=True)

# Ініціалізація WBT
wbt = WhiteboxTools()
wbt.set_working_dir(str(output_folder))
wbt.verbose = True

# DEM файли
dem_files = sorted(dem_folder.glob("*_breached.tif"))

# 🔁 Обхід DEM
for dem_path in dem_files:
    name = dem_path.stem.replace("_breached", "")  # наприклад tan_dem

    # Вихідні шляхи
    slope_raster = output_folder / f"{name}_slope.tif"
    pointer_raster = output_folder / f"{name}_d8_pointer.tif"
    accum_raster = output_folder / f"{name}_d8_accum.tif"
    twi_raster = output_folder / f"{name}_twi.tif"

    print(f"\n🌄 Обчислюється slope для {name}")
    wbt.slope(
        dem=str(dem_path),
        output=str(slope_raster),
        zfactor=1.0
    )

    print(f"🧭 Обчислюється D8 Pointer для {name}")
    wbt.d8_pointer(
        dem=str(dem_path),
        output=str(pointer_raster)
    )

    print(f"💧 Обчислюється D8 Accumulation для {name}")
    wbt.run_tool(
        "D8FlowAccumulation",
        [
            f"--dem={str(dem_path)}",
            f"--output={str(accum_raster)}",
            "--out_type=specific contributing area"
        ]
    )


    print(f"🌊 Обчислюється TWI для {name}")
    wbt.wetness_index(
        sca=str(accum_raster),
        slope=str(slope_raster),
        output=str(twi_raster)
    )

    print(f"✅ Завершено: {name}")

print(f"\n🎉 Усі TWI та інші продукти збережено в: {output_folder}")
```
