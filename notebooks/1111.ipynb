```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()

G.add_edges_from([
    ("i", "j"),
    ("j", "k"),
    ("k", "m"),
    ("i", "s"),
    ("s", "k"),
])

pos = nx.spring_layout(G, seed=42)

plt.figure(figsize=(6,4))
nx.draw(G, pos, with_labels=True, node_size=1500,
        node_color="#d9eaff", arrowsize=20,
        font_size=12, font_weight="bold")
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()

# Nodes
G.add_nodes_from(["i", "j", "k", "m", "s"])

# Edges
G.add_edge("i", "j")
G.add_edge("j", "k")
G.add_edge("k", "m")
G.add_edge("i", "s")
G.add_edge("s", "k")

# Positioning similar to your TikZ layout
pos = {
    "i": (0, 0),
    "j": (2, 1),
    "k": (4, 0),
    "m": (6, 1),
    "s": (2, -1.4)
}

plt.figure(figsize=(6,4))
nx.draw(
    G, pos,
    with_labels=True,
    arrows=True,
    node_size=1500,
    node_color="#FFFFFF",
    edge_color="#333333",
    font_size=12,
    linewidths=1,
    width=1.6,
    arrowstyle="->",
    arrowsize=15
)

plt.axis("off")
plt.tight_layout()
plt.savefig("causal_graph.png", dpi=300)
plt.show()
```
