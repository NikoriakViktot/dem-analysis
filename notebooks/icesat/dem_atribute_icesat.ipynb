```python
import rasterio
import geopandas as gpd
from rasterio.sample import sample_gen
import numpy as np
from pathlib import Path
```

```python
attr_folder = Path("/mnt/c/Users/5302/OneDrive/PhD/paper_DEM_artickle/data//DATA_UTM/dem/terrain_attributes")
```

```python
attr_paths = sorted(attr_folder.glob("*.tif"))
attr_paths
```

```python
ice_gdf = gpd.read_parquet("/mnt/c/Users/5302/OneDrive/PhD/paper_DEM_artickle/data/data_icesat2/icesat2_with_lulc.parquet")
```

```python
coords = [(geom.x, geom.y) for geom in ice_gdf.geometry]
```

```python

# ✅ Пройтись по всіх атрибутах
for path in attr_paths:
    filename = path.stem  # napриклад: tan_dem_utm32635_tri
    parts = filename.split("_")
    # Витягуємо назву DEM + атрибут
    dem_name = "_".join(parts[:2])  # tan_dem
    attribute = parts[-1]           # tri
    col_name = f"{dem_name}_{attribute}"

    # Витяг значень
    with rasterio.open(path) as src:
        if not src.crs == ice_gdf.crs:
            raise ValueError(f"❌ CRS не збігаються у {col_name}")
        sampled = list(src.sample(coords))

    # Запис значень у колонку
    values = [val[0] if val and val[0] != src.nodata else np.nan for val in sampled]
    ice_gdf[col_name] = values

    print(f"✅ Додано: {col_name}")

# 💾 Збереження
ice_gdf.to_parquet("/mnt/c/Users/5302/OneDrive/PhD/paper_DEM_artickle/data/data_icesat2/icesat2_with_terrain_attributes.parquet")
print("📦 Готово: всі атрибути записані в файл")
```
