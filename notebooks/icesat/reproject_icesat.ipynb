```python
# 📖 ICESat дані
tracks = gpd.read_parquet("/mnt/c/Users/5302/PycharmProjects/GeoHydroAI/data/icesat2_dem.parquet")
print(tracks)
```

```python
basin_4326 = reader("data_raw/basin_bil_cher_buf_500.gpkg").read_gpkg()

print(basin_4326)
```

```python
# 1. Об'єднуємо всі полігони у один (це весь басейн!)
basin_poly = basin_4326.geometry.union_all()
print(basin_poly.bounds)
```

```python
if tracks.crs != basin_4326.crs:
    # Приводимо GeoDataFrame до потрібної CRS
    basin = basin_4326.to_crs(tracks.crs)
```

```python
tracks_in_basin = tracks[tracks.within(basin_poly)].copy()
tracks_in_basin["time"] = tracks_in_basin.index
```

```python
tracks_proj = tracks_in_basin.to_crs(epsg=4326)
tracks_proj.to_parquet("/mnt/c/Users/5302/OneDrive/PhD/paper_DEM_artickle/data/data_icesat2/night_cut_4326.parquet")
```
