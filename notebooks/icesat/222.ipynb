```python
import os, math, time
import duckdb, pandas as pd, numpy as np, rasterio
from rasterio import sample as rio_sample
import pyarrow as pa
import pyarrow.parquet as pq

# --- шляхи
PARQUET_PATH = "/mnt/c/Users/5302/PycharmProjects/GeoHydroAI-V2/data/parquet/NMAD_dem.parquet"
RASTER_PATH  = r"/mnt/d/geoembeding/2024/geo_embeddings_2024_64ch.tif"
OUT_PARQUET = "/mnt/d/geoembeding/NMAD_with_embeddings_track_2.parquet"
os.makedirs(os.path.dirname(OUT_PARQUET), exist_ok=True)

# --- параметри
BATCH = 500_000  # під RAM

# --- колонки з твого датасету
SELECT_COLS = ["rgt", "track", "spot",
               "region", "sc_orient", "segment_dist", "solar_elevation", "segment_id",
               "background_rate", "cycle", "pair", "snowcover", "yapc_score", "y_atc",
               "landcover", "x_atc", "quality_ph", "time",
               "height", "geoid_height", "orthometric_height",
               "h_alos_dem", "delta_alos_dem", "abs_delta_alos_dem",
               "h_aster_dem", "delta_aster_dem", "abs_delta_aster_dem",
               "h_copernicus_dem", "delta_copernicus_dem", "abs_delta_copernicus_dem",
               "h_fab_dem", "delta_fab_dem", "abs_delta_fab_dem",
               "h_nasa_dem", "delta_nasa_dem", "abs_delta_nasa_dem",
               "h_srtm_dem", "delta_srtm_dem", "abs_delta_srtm_dem",
               "h_tan_dem", "delta_tan_dem", "abs_delta_tan_dem",
               "lulc_class", "lulc_name",
               "alos_dem_slope", "aster_dem_slope", "copernicus_dem_slope",
               "fab_dem_slope", "nasa_dem_slope", "srtm_dem_slope", "tan_dem_slope",
               "alos_dem_geomorphon", "alos_dem_landform",
               "aster_dem_geomorphon", "aster_dem_landform",
               "copernicus_dem_geomorphon", "copernicus_dem_landform",
               "fab_dem_geomorphon", "fab_dem_landform",
               "nasa_dem_geomorphon", "nasa_dem_landform",
               "srtm_dem_geomorphon", "srtm_dem_landform",
               "tan_dem_geomorphon", "tan_dem_landform",
               "alos_dem_2000", "aster_dem_2000", "copernicus_dem_2000",
               "fab_dem_2000", "nasa_dem_2000", "srtm_dem_2000", "tan_dem_2000",
               "alos_dem_stream", "aster_dem_stream", "copernicus_dem_stream",
               "fab_dem_stream", "nasa_dem_stream", "srtm_dem_stream", "tan_dem_stream",
               "alos_dem_twi", "aster_dem_twi", "copernicus_dem_twi",
               "fab_dem_twi", "nasa_dem_twi", "srtm_dem_twi", "tan_dem_twi",
               "x", "y", "x_merc", "y_merc",
               "nmad_alos", "nmad_aster", "nmad_cop", "nmad_fab", "nmad_nasa", "nmad_srtm", "nmad_tan"]

# --- підготовка
con = duckdb.connect()
tbl = f"read_parquet('{PARQUET_PATH}')"

with rasterio.open(RASTER_PATH) as src:
    assert src.count == 64, f"expected 64 bands, got {src.count}"
    print("Raster CRS:", src.crs)
    print("Raster bounds:", src.bounds)

    n_rows = con.execute(f"SELECT count(*) FROM {tbl}").fetchone()[0]
    n_batches = math.ceil(n_rows / BATCH)
    print(f"Total rows: {n_rows:,} | Batch size: {BATCH:,} | Batches: {n_batches}")

    writer = None
    t0 = time.time()

    for bi, offset in enumerate(range(0, n_rows, BATCH), start=1):
        qcols = ", ".join(SELECT_COLS)
        df = con.execute(
            f"SELECT {qcols} FROM {tbl} LIMIT {BATCH} OFFSET {offset}"
        ).fetchdf()

        coords = list(zip(df["x"].to_numpy(), df["y"].to_numpy()))

        # вибірка 64-канальних значень; masked -> NaN
        vals = []
        for v in rio_sample.sample_gen(src, coords):
            a = np.asarray(v, dtype="float32")
            if np.ma.isMaskedArray(v):
                a = np.where(v.mask, np.nan, a)
            vals.append(a)

        emb_mat = np.vstack(vals)  # shape (batch, 64)
        emb_cols = [f"emb_{i:03d}" for i in range(1, src.count + 1)]
        emb = pd.DataFrame(emb_mat, columns=emb_cols)

        # додаткова одна колонка зі всім вектором (список на 64 елементи)
        emb["emb_all"] = emb_mat.tolist()

        out = pd.concat([df.reset_index(drop=True), emb], axis=1)

        # запис у Parquet батчем
        table = pa.Table.from_pandas(out, preserve_index=False)
        if writer is None:
            writer = pq.ParquetWriter(
                OUT_PARQUET, table.schema,
                compression="zstd",
                use_dictionary=True
            )
        writer.write_table(table)

        # прогрес
        done = min(offset + BATCH, n_rows)
        elapsed = time.time() - t0
        speed = done / max(elapsed, 1e-6)
        eta = (n_rows - done) / max(speed, 1e-6)
        print(f"[{bi:03d}/{n_batches:03d}] wrote rows: {done:,}/{n_rows:,} "
              f"({done/n_rows:5.1%}) | {speed:,.0f} rows/s | ETA ~ {eta/60:,.1f} min")

    if writer is not None:
        writer.close()

print("DONE ->", os.path.abspath(OUT_PARQUET))
```
