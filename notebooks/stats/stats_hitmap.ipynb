```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# Input data (RMSE, m) @ GSD=30 m
# -----------------------------
dems = ["ALOS", "ASTER", "Copernicus", "FAB", "NASA", "SRTM", "TAN"]
# Slope classes (INSPIRE)
classes = [
    ("Flat (0–2°)", "flat"),
    ("Undulating (2–6°)", "undulating"),
    ("Hilly (6–12°)", "hilly"),
    ("Mountainous (> 12°)", "mountainous"),
]
# RMSE values per class (order of DEMs as in `dems`)
rmse = {
    "flat":        [3.11, 10.62, 3.33, 3.38, 3.64, 4.25, 3.22],
    "undulating":  [6.43, 11.51, 5.66, 3.97, 6.29, 5.81, 5.51],
    "hilly":       [8.21, 12.09, 6.90, 4.57, 8.11, 6.97, 6.93],
    "mountainous": [12.95, 15.38, 12.46, 7.43, 12.46, 10.21, 12.45],
}

GSD = 30.0

# INSPIRE Rec.52 bands for RMSE at GSD=30 m
# flat/undulating: RMSE in [GSD/20, GSD/3] = [1.5, 10]
# hilly/mountainous: RMSE in [GSD/10, GSD/3] = [3, 10]
bands_rmse = {
    "flat":        (GSD/20.0, GSD/3.0),   # [1.5, 10]
    "undulating":  (GSD/20.0, GSD/3.0),   # [1.5, 10]
    "hilly":       (GSD/10.0, GSD/3.0),   # [3, 10]
    "mountainous": (GSD/10.0, GSD/3.0),   # [3, 10]
}

# Corresponding bands for the ratio R = GSD/RMSE
# flat/undulating: RMSE ∈ [GSD/20, GSD/3]  => R ∈ [3, 20]
# hilly/mountainous: RMSE ∈ [GSD/10, GSD/3] => R ∈ [3, 10]
bands_ratio = {
    "flat":        (3.0, 20.0),
    "undulating":  (3.0, 20.0),
    "hilly":       (3.0, 10.0),
    "mountainous": (3.0, 10.0),
}

# -----------------------------
# Figure 1: RMSE with compliant band per slope class
# -----------------------------
fig, axes = plt.subplots(2, 2, figsize=(12, 8), constrained_layout=True)
axes = axes.ravel()

for ax, (title, key) in zip(axes, classes):
    y = rmse[key]
    x = np.arange(len(dems))
    band_lo, band_hi = bands_rmse[key]

    # shaded compliant band
    ax.axhspan(band_lo, band_hi, alpha=0.15, lw=0, label="INSPIRE-compliant band")

    # plot DEM points; color non-compliant in open markers
    for i, val in enumerate(y):
        compliant = (band_lo <= val <= band_hi)
        ax.plot(i, val,
                marker="o",
                markersize=7,
                linestyle="none",
                markerfacecolor=("white" if not compliant else None),
                markeredgecolor="black")

    # cosmetics
    ax.set_title(f"RMSE by DEM — {title}", fontsize=12)
    ax.set_xticks(x)
    ax.set_xticklabels(dems, rotation=30, ha="right")
    ax.set_ylabel("RMSE (m)")
    # y-limits: nice headroom
    ymax = max(max(y), band_hi) * 1.15
    ymin = 0.0
    ax.set_ylim(ymin, ymax)
    ax.grid(True, axis="y", linestyle=":", linewidth=0.6, alpha=0.7)

# Single legend
handles, labels = axes[0].get_legend_handles_labels()
if handles:
    fig.legend(handles, labels, loc="upper center", ncol=1, frameon=False)

# fig.suptitle("Root Mean Square Error (RMSE) vs INSPIRE Rec.52 Bands (GSD = 30 m)", fontsize=14, y=1.02)
plt.show()

# -----------------------------
# Figure 2 (optional): Ratio GSD/RMSE with compliant band
# -----------------------------
fig2, axes2 = plt.subplots(2, 2, figsize=(12, 8), constrained_layout=True)
axes2 = axes2.ravel()

for ax, (title, key) in zip(axes2, classes):
    r = [GSD / v for v in rmse[key]]
    x = np.arange(len(dems))
    band_lo, band_hi = bands_ratio[key]

    # shaded compliant band
    ax.axhspan(band_lo, band_hi, alpha=0.15, lw=0, label="INSPIRE-compliant band")

    # plot DEM points; open marker if non-compliant
    for i, val in enumerate(r):
        compliant = (band_lo <= val <= band_hi)
        ax.plot(i, val,
                marker="s",
                markersize=7,
                linestyle="none",
                markerfacecolor=("white" if not compliant else None),
                markeredgecolor="black")

    ax.set_title(f"GSD/RMSE by DEM — {title}", fontsize=12)
    ax.set_xticks(x)
    ax.set_xticklabels(dems, rotation=30, ha="right")
    ax.set_ylabel("GSD / RMSE (unitless)")
    ymax = max(max(r), band_hi) * 1.15
    ymin = 0.0
    ax.set_ylim(ymin, ymax)
    ax.grid(True, axis="y", linestyle=":", linewidth=0.6, alpha=0.7)

handles2, labels2 = axes2[0].get_legend_handles_labels()
if handles2:
    fig2.legend(handles2, labels2, loc="upper center", ncol=1, frameon=False)

# fig2.suptitle("Scale Consistency (GSD/RMSE) vs INSPIRE Rec.52 Bands (GSD = 30 m)", fontsize=14, y=1.02)
plt.show()
```

```python
# Floodplain (HAND 0–6 m): compact table + tighter bar chart
# Run locally: python 3.9+, pandas, matplotlib, numpy

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ---- Data ----
dems = ["ALOS", "ASTER", "Copernicus", "FABDEM", "NASA", "SRTM", "TAN DEM"]
rmse = [8.42, 10.50, 6.54, 5.05, 10.14, 8.10, 6.55]  # meters

# constants
rise = 6.0             # critical rise (m)
LAWA = 0.20            # m
FEMA = 0.125           # m
INSPIRE_upper = 10.0   # m (flat/undulating upper bound at GSD=30 m)

# ---- Table (CSV) ----
df = pd.DataFrame({
    "DEM": dems,
    "RMSE (m)": rmse,
    "RMSE / 6 m (%)": [round(v / rise * 100, 1) for v in rmse],
    "LAWA × (RMSE/0.20)": [round(v / LAWA, 2) for v in rmse],
    "FEMA × (RMSE/0.125)": [round(v / FEMA, 2) for v in rmse],
    "INSPIRE (≤10 m)": ["✅" if v <= INSPIRE_upper else "❌" for v in rmse],
}).sort_values("RMSE (m)")

df.to_csv("floodplain_metrics.csv", index=False)
print("Saved table -> floodplain_metrics.csv")
print(df)

# ---- Compact bar chart ----
fig, ax = plt.subplots(figsize=(8, 4))  # compact

x = np.arange(len(dems))
bars = ax.bar(x, rmse, width=0.7, edgecolor="black", linewidth=0.6, color="#9aa0a6")  # neutral gray

# Standards (colored lines)
ax.axhline(INSPIRE_upper, color="green", linestyle="--", linewidth=1.2, label="INSPIRE upper (10 m)")
ax.axhline(LAWA,           color="blue",  linestyle="--", linewidth=1.2, label="LAWA (0.20 m)")
ax.axhline(FEMA,           color="red",   linestyle="--", linewidth=1.2, label="FEMA High/Flat (0.125 m)")

# Axis labels & ticks
ax.set_ylabel("RMSE (m)")
ax.set_xticks(x)
ax.set_xticklabels(dems, rotation=30, ha="right")
ax.set_title("Floodplain (HAND 0–6 m): DEM RMSE vs Regulatory Standards")

# Percent labels above bars (RMSE / 6 m)
ymax = max(rmse) * 1.15
ax.set_ylim(0, ymax)
pct = [v / rise * 100 for v in rmse]
for rect, p in zip(bars, pct):
    y = rect.get_height()
    y_text = min(y + 0.25, ymax - 0.3)
    ax.text(rect.get_x() + rect.get_width()/2, y_text, f"{p:.0f}%",
            ha="center", va="bottom", fontsize=8)

ax.legend(loc="upper right", fontsize=8, frameon=False)

plt.tight_layout()
plt.savefig("floodplain_rmse_vs_standards.png", dpi=300, bbox_inches="tight")
plt.show()

print("Saved figure -> floodplain_rmse_vs_standards.png")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ---- Data ----
dems = ["ALOS", "ASTER", "Copernicus", "FABDEM", "NASA", "SRTM", "TAN DEM"]
rmse = [8.42, 10.50, 6.54, 5.05, 10.14, 8.10, 6.55]

rise = 6.0
LAWA = 0.20
FEMA = 0.125
INSPIRE_upper = 10.0

# ---- Table ----
df = pd.DataFrame({
    "DEM": dems,
    "RMSE (m)": rmse,
    "RMSE / 6 m (%)": [round(v / rise * 100, 1) for v in rmse],
    "LAWA × (RMSE/0.20)": [round(v / LAWA, 2) for v in rmse],
    "FEMA × (RMSE/0.125)": [round(v / FEMA, 2) for v in rmse],
    "INSPIRE (≤10 m)": ["✅" if v <= INSPIRE_upper else "❌" for v in rmse],
})
df.to_csv("floodplain_metrics.csv", index=False)

# ---- Plot ----
fig, ax = plt.subplots(figsize=(7, 4))
x = np.arange(len(dems))

# highlight FABDEM and TAN DEM
colors = []
for name in dems:
    if "FAB" in name or "TAN" in name:
        colors.append("#4a90e2")  # синюватий для виділення
    else:
        colors.append("#c0c0c0")  # світло-сірий

bars = ax.bar(x, rmse, width=0.65, edgecolor="black", color=colors)

# Standards
ax.axhline(INSPIRE_upper, color="green", linestyle="--", linewidth=1.2, label="INSPIRE upper (10 m)")
ax.axhline(LAWA, color="blue", linestyle="--", linewidth=1.2, label="LAWA (0.20 m)")
ax.axhline(FEMA, color="red", linestyle="--", linewidth=1.2, label="FEMA High/Flat (0.125 m)")

# Axis & labels
ax.set_ylabel("RMSE (m)", fontsize=11)
ax.set_xticks(x)
ax.set_xticklabels(dems, rotation=30, ha="right", fontsize=10)
ax.set_ylim(0, max(rmse) * 1.2)
ax.set_title("Floodplain (HAND 0–6 m): DEM RMSE vs Regulatory Standards", fontsize=12)

# Add % labels above bars
for rect, val in zip(bars, rmse):
    ax.text(rect.get_x() + rect.get_width()/2, rect.get_height()+0.2,
            f"{val/rise*100:.0f}%", ha="center", va="bottom", fontsize=8)

# Legend
ax.legend(loc="upper right", fontsize=9, frameon=False)

plt.tight_layout()
plt.savefig("floodplain_rmse_vs_standards.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# Data (Floodplain: HAND 0–6 m)
# -----------------------------
dems  = ["ALOS", "ASTER", "Copernicus", "FABDEM", "NASA", "SRTM", "TanDEM-X"]
rmse  = [8.42,   10.50,    6.54,        5.05,     10.14,  8.10,   6.55]  # meters

# Reference thresholds
RISE = 6.0          # critical flood-rise band (HAND 0–6 m)
LAWA = 0.20         # m
FEMA = 0.125        # m
INSPIRE_UPPER = 10  # m (flat/undulating at GSD=30 m)

# Table (optional): save alongside the figure
df = pd.DataFrame({
    "DEM": dems,
    "RMSE (m)": rmse,
    "RMSE / 6 m (%)": [round(v/RISE*100, 1) for v in rmse],
    "LAWA × (RMSE/0.20)": [round(v/LAWA, 2) for v in rmse],
    "FEMA × (RMSE/0.125)": [round(v/FEMA, 2) for v in rmse],
    "INSPIRE (≤10 m)": ["✅" if v <= INSPIRE_UPPER else "❌" for v in rmse],
})
df.to_csv("floodplain_metrics.csv", index=False)

# -----------------------------
# Figure
# -----------------------------
plt.rcParams.update({
    "figure.dpi": 150,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.linewidth": 0.8,
})

fig, ax = plt.subplots(figsize=(7.6, 4.6))

x = np.arange(len(dems))

# Subtle academic style bars:
#   - default neutral gray
#   - highlight FABDEM and TAN DEM
colors = ["#bcbec4"] * len(dems)
for i, name in enumerate(dems):
    if "FAB" in name:
        colors[i] = "#587ba8"  # highlight FABDEM


bars = ax.bar(x, rmse, width=0.66, edgecolor="black", linewidth=0.7, color=colors)

# Regulatory/reference thresholds (explicit colors)
ax.axhline(INSPIRE_UPPER, color="green", linestyle="--", linewidth=1.2, label="INSPIRE upper (10 m)")
ax.axhline(LAWA,          color="blue",  linestyle="--", linewidth=1.2, label="LAWA (0.20 m)")
ax.axhline(FEMA,          color="red",   linestyle="--", linewidth=1.2, label="FEMA High/Flat (0.125 m)")

# Axes
ax.set_ylabel("RMSE (m)")
ax.set_xticks(x)
ax.set_xticklabels(dems, rotation=28, ha="right")
ax.set_ylim(0, max(rmse)*1.20)
# ax.set_title("Floodplain (HAND 0–6 m): DEM RMSE vs Regulatory Thresholds")

# Percentage labels = share of the 6 m flood-rise band
pct = [v/RISE*100 for v in rmse]
ymax = ax.get_ylim()[1]
for rect, p in zip(bars, pct):
    y = rect.get_height()
    y_text = min(y + 0.25, ymax - 0.35)  # keep inside the frame
    ax.text(rect.get_x() + rect.get_width()/2, y_text,
            f"{p:.0f}% of 6 m rise", ha="center", va="bottom", fontsize=8)

# Legend outside the axes
leg = ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.14), ncol=3, frameon=False)

plt.tight_layout()
plt.savefig("floodplain_rmse_vs_thresholds.png", dpi=300, bbox_inches="tight")
plt.savefig("floodplain_rmse_vs_thresholds.pdf", dpi=300, bbox_inches="tight")
plt.show()

print("Saved: floodplain_rmse_vs_thresholds.png / .pdf")
print("Saved table: floodplain_metrics.csv")
```
