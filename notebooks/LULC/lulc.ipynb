```python
lulc_array = read_dem(path="LULC/lulc2018.tif", mode="array")

print(lulc_array)
```

```python
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import numpy as np

# Твій DataArray (наприклад, після reader.read_raster("array"))
lulc = lulc_array  # наприклад: lulc = reader.read_raster("array")

# Словник: клас → назва → колір
lulc_classes_colors = {
    0:  {"name": "No Data", "color": "#FFFFFF"},
    1:  {"name": "Water", "color": "#1A5BAB"},
    2:  {"name": "Trees", "color": "#358221"},
    4:  {"name": "Flooded Vegetation", "color": "#87D19E"},
    5:  {"name": "Crops", "color": "#FFDB5C"},
    7:  {"name": "Built Area", "color": "#ED022A"},
    8:  {"name": "Bare Ground", "color": "#EDE9E4"},
    9:  {"name": "Snow/Ice", "color": "#F2FAFF"},
    10: {"name": "Clouds", "color": "#C8C8C8"},
    11: {"name": "Rangeland", "color": "#A7D282"}
}


# Підготовка до візуалізації
labels = list(lulc_classes_colors.keys())
colors = [lulc_classes_colors[key]["color"] for key in labels]
names = [lulc_classes_colors[key]["name"] for key in labels]

# Класифікаційна палітра
cmap = mcolors.ListedColormap(colors)
bounds = np.array(labels + [max(labels) + 1]) - 0.5
norm = mcolors.BoundaryNorm(bounds, cmap.N)

# Побудова карти
plt.figure(figsize=(12, 8))
im = plt.imshow(lulc.values, cmap=cmap, norm=norm)
plt.title("LULC Classification Map")
plt.xlabel("x")
plt.ylabel("y")

# Створення легенди
patches = [plt.plot([],[], marker="s", ms=10, ls="", mec=None, color=cmap(i))[0] for i in range(len(labels))]
plt.legend(patches, names, loc='center left', bbox_to_anchor=(1, 0.5), title="Land Cover Classes")

plt.tight_layout()
plt.show()

```

```python
import numpy as np

# 1. Знаходимо унікальні класи в lulc (масив може бути np.ndarray чи xarray.DataArray)
unique_classes = np.unique(lulc.values if hasattr(lulc, "values") else lulc)

# 2. Відбираємо тільки ті, що є у lulc_classes_colors
present_classes = [c for c in unique_classes if c in lulc_classes_colors]

# 3. Формуємо списки для візуалізації
colors_present = [lulc_classes_colors[c]["color"] for c in present_classes]
names_present = [lulc_classes_colors[c]["name"] for c in present_classes]

# 4. Оновлений colormap і нормалізація тільки для присутніх класів
cmap_present = mcolors.ListedColormap(colors_present)
bounds_present = np.array(present_classes + [max(present_classes) + 1]) - 0.5
norm_present = mcolors.BoundaryNorm(bounds_present, cmap_present.N)

# 5. Побудова карти (змінюємо cmap та norm)
plt.figure(figsize=(12, 8))
im = plt.imshow(lulc.values if hasattr(lulc, "values") else lulc, cmap=cmap_present, norm=norm_present)
plt.title("LULC Classification Map")
plt.xlabel("x")
plt.ylabel("y")

# 6. Легенда тільки для присутніх класів
patches = [plt.plot([],[], marker="s", ms=10, ls="", mec=None, color=cmap_present(i))[0] for i in range(len(present_classes))]
plt.legend(patches, names_present, loc='center left', bbox_to_anchor=(1, 0.5), title="Land Cover Classes")

plt.tight_layout()
plt.show()
```
