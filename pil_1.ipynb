```python
from PIL import Image
from pathlib import Path

BASE_DIR = Path("/mnt/c/Users/5302/OneDrive/PhD/paper_DEM_artickle/text/Figure_img")
TARGET_WIDTH_PX = 3800
DPI = (300, 300)

files = {
    "Fig_1.jpg": "Fig1_StudyArea.jpg",
    "Figure_3_10.jpg": "Fig3_10_HAND_5m_FloodZones.jpg",
}

for src_name, out_name in files.items():
    src_path = BASE_DIR / src_name
    out_path = BASE_DIR / out_name

    print(f"\nProcessing: {src_path.name}")
    print("Exists:", src_path.exists())

    img = Image.open(src_path).convert("RGB")

    new_height = int(TARGET_WIDTH_PX * img.height / img.width)
    img = img.resize((TARGET_WIDTH_PX, new_height), Image.LANCZOS)

    img.save(
        out_path,
        format="JPEG",
        dpi=DPI,
        quality=95,
        subsampling=0
    )

    size_mb = out_path.stat().st_size / 1024 / 1024
    print(f"Saved → {out_path.name} | size = {size_mb:.2f} MB")

```
